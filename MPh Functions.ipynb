### Build the geometry:
> model.build()


### Build the mesh:
>model.mesh()

### Solve a study:
> model.solve()
>
> model.solve('')


### Evaluate:
> model.evaluate('variable_a', 'variable_b')
> - Returns NumPy arrays
> - Example:
>> (x, y, E) = model.evaluate(['x', 'y', 'es.normE'])


### Exports:
> model.exports()
>
> model.export('image', 'static field.png')


### Save:
> model.save('model_name')

### Compute:
- Build
> build(geometry=None)
- Mesh
> mesh(mesh=None)
- Solve
> solve(study=None)
- Evaluate
> evalute(expression, unit=None, dataset=None, inner=None, outer=None)
